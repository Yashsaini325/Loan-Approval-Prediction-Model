```python
# Converting Data into tensor
import torch
from torch import nn
from torch.utils.data import DataLoader, TensorDataset
import torch.optim as optim

# Training data
X_train_tensor = torch.tensor(X_train_scaled, dtype=torch.float32)
y_train_tensor = torch.tensor(y_train.values, dtype=torch.float32).view(-1,1)

# Testing Data
X_test_tensor = torch.tensor(X_test_scaled, dtype=torch.float32)
y_test_tensor = torch.tensor(y_test.values, dtype=torch.float32).view(-1,1)
```

```python
print(X_train_tensor.shape[0], y_train_tensor.shape[0])  # Must be equal
print(X_test_tensor.shape[0], y_test_tensor.shape[0])    # Must be equal
```

```python
# Accessing Tensor from RAM
train_dataset = TensorDataset(X_train_tensor, y_train_tensor)
test_dataset = TensorDataset(X_test_tensor, y_test_tensor)
```

```python
# Creating Batches
train_loader = DataLoader(train_dataset, batch_size=32)
test_loader = DataLoader(test_dataset, batch_size=32)
```

```python
# ANN (Binary classification)
class ANN(nn.Module):
    def __init__(self):
        super(ANN, self).__init__()

        self.model = nn.Sequential(

           # 1st Hidden Layer
            nn.Linear(X_train_tensor.shape[1], 32),
            nn.ReLU(),

            # # 2nd Hidden Layer
            # nn.Linear(32, 32),
            # nn.ReLU(),

            # output layer
            nn.Linear(32, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        x = self.model(x)
        return x
        
```

```python
# creating Object
model = ANN()
```

```python
# Loss function nad Optimizers
loss = nn.MSELoss()
optimizer = optim.Adam(model.parameters())
```

```python
# Training ANN

epochs = 25

for epoch in range(epochs):
    running_loss = 0.0
    # Put model in training mode
    model.train()

    for xb, yb in train_loader:

        optimizer.zero_grad()
        output = model(xb)
        error = loss(output, yb)
        error.backward()
        optimizer.step()
    
        running_loss += error.item()

    epoch_training_loss = running_loss / len(train_loader)



    # Validation
    # Put model in evaluation mode
    model.eval()
    running_val_loss = 0.0

    with torch.no_grad():
        for xb, yb in test_loader:
            output = model(xb)
            error = loss(output, yb)
            running_val_loss += error.item()

        epoch_eval_loss = running_val_loss / len(test_loader)
        
    print(f"Epoch: {epoch+1}, training_loss: {epoch_training_loss}, eval_loss = {epoch_eval_loss}")
```

```python
# evaluation

from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score, confusion_matrix, classification_report

# Ensure evaluation mode without gradients
model.eval()
with torch.no_grad():
    # Model predictions
    y_pred = (model(X_test_tensor) >= 0.5).float().cpu().numpy()
    train_predict = (model(X_train_tensor) >= 0.5).float().cpu().numpy()

# Convert target tensors to numpy arrays
y_test_np = y_test_tensor.cpu().numpy()
y_train_np = y_train_tensor.cpu().numpy()

# Print metrics
print(f"precision_score: {precision_score(y_test_np, y_pred)}")
print(f"recall_score: {recall_score(y_test_np, y_pred)}")
print(f"f1_score: {f1_score(y_test_np, y_pred)}")
print(f"training accuracy_score: {accuracy_score(y_train_np, train_predict)}")
print(f"Testing accuracy_score: {accuracy_score(y_test_np, y_pred)}\n")
print(f"CM: {confusion_matrix(y_test_np, y_pred)}\n")
print(f"CR: {classification_report(y_test_np, y_pred)}\n")
```
